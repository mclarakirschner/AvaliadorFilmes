3. Exemplo de função com argumento

```python
def cumprimentar(cliente,setor):
  print(f"Seja bem-vindo, {cliente}!")
  print(f"Ativar setor {setor}!")
```

```python
cumprimentar("Ana Julia","TI")
cumprimentar("João","TI")
cumprimentar(2027, "Calendário")
cumprimentar("","")
```

4. Exemplo função **retorno**

```python
def pre_processar(frase):
  tokens = frase.lower().split()
  return tokens
```

```python
tokens_retornados = pre_processar("Estou revisando funções em Python")
print(tokens_retornados)
```

```python
def pre_processar_contar(frase):
  tokens = frase.lower().split()
  qtd_tokens = len(tokens)
  return tokens
```

```python
tokens_retornados, qtd_tokens = pre_processar_contar("Olá mundo")
print(tokens_retornados)
print(2+2)
print(qtd_tokens)
```
